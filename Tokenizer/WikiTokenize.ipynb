```python
from datasets import load_dataset

ds = load_dataset("Salesforce/wikitext", "wikitext-103-raw-v1")
```

```python
import os
output_dir = "..\Data"
output_file = os.path.join(output_dir, "wikitext103_train.txt")

# Ensure the directory path exists
os.makedirs(output_dir, exist_ok=True)

# 3. Stream the text rows out to your target file
print(f"Writing dataset lines out to: {output_file}...")
with open(output_file, "w", encoding="utf-8") as f:
    for item in ds["train"]:
        text = item["text"]
        f.write(text + "\n")

print("✨ Success! Your 'wikitex")
```
