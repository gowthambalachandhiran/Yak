```python

import os
import pandas as pd
from datasets import load_dataset

# Download SNLI from Hugging Face
dataset = load_dataset("stanfordnlp/snli")

# Check the notebook's working directory
output_dir = os.getcwd()
print("Saving files to:", output_dir)

# Convert each split to CSV
for split in ["train", "validation", "test"]:
    df = dataset[split].to_pandas()

    # Keep the three main columns
    df = df[["label", "premise", "hypothesis"]]

    # Remove examples without valid labels
    df = df[df["label"].isin([0, 1, 2])].copy()

    # Convert numeric labels to readable names
    label_map = {
        0: "entailment",
        1: "neutral",
        2: "contradiction"
    }
    df["label"] = df["label"].map(label_map)

    # Save beside the notebook
    filename = f"snli_{split}.csv"
    df.to_csv(filename, index=False)

    print(f"Saved {filename}: {len(df):,} rows")

print("Download and conversion complete!")
```
