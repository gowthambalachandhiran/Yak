```python
import os
output_dir = r".\Data\processed"
os.makedirs(output_dir, exist_ok=True)
```

```python
print("Loading dataset from cache...")
# Replace 'snli' with whichever dataset name your script processed (e.g., 'banking77', 'clinc_oos', etc.)
dataset = load_dataset("stanfordnlp/snli") 

print(f"Saving dataset to {output_dir}...")
dataset.save_to_disk(output_dir)
```

```python
# 2. Define the path inside the Yak repository
from pathlib import Path
repo_dir = next((p for p in (Path.cwd(), *Path.cwd().parents) if (p / "Data" / "snli.ipynb").is_file()), Path.cwd())
output_dir = str(repo_dir / "Data" / "snli_processed")
os.makedirs(output_dir, exist_ok=True)

# 3. Save the processed dataset to disk
print(f"Saving dataset to {output_dir}...")
dataset.save_to_disk(output_dir)
```

```python
data_dir = r".\Data"
# Loop through the splits in your loaded 'dataset' object (train, validation/dev, test)
for split_name, split_data in dataset.items():
    # Convert the Hugging Face dataset split to a Pandas DataFrame
    df = split_data.to_pandas()
    
    # Define the CSV file path
    csv_path = os.path.join(data_dir, f"snli_{split_name}.csv")
    
    # Save to CSV (index=False avoids saving row numbers)
    df.to_csv(csv_path, index=False, encoding='utf-8')
    print(f"Saved: {csv_path} ({len(df):,} rows)")

print("All splits successfully converted to CSV in your Data folder!")
```
