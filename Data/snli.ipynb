```python
data_dir = r".\Data"
# Loop through the splits in your loaded 'dataset' object (train, validation/dev, test)
for split_name, split_data in dataset.items():
    # Convert the Hugging Face dataset split to a Pandas DataFrame
    df = split_data.to_pandas()
    
    # Define the CSV file path
    csv_path = os.path.join(data_dir, f"snli_{split_name}.csv")
    
    # Save to CSV (index=False avoids saving row numbers)
    df.to_csv(csv_path, index=False, encoding='utf-8')
    print(f"Saved: {csv_path} ({len(df):,} rows)")

print("All splits successfully converted to CSV in your Data folder!")
```
